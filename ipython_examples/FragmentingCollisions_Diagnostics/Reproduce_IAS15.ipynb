```python
import rebound
import reboundx
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

pd.set_option('display.precision', 20)
```

```python
heartbeat_column_names = [
    "t", "walltime", "N", "total_mass", 
    "ang_mom_x", "ang_mom_y", "ang_mom_z", "energy", 
    "p3_m", "p3_r", "p3_x", "p3_y", "p3_z", 
    "p3_vx", "p3_vy", "p3_vz"
]

# Read the files
heartbeat_embryo_3 = pd.read_csv(
    "diagnose_TRACE/heartbeat_particle_3_2_IAS15.txt", 
    names=heartbeat_column_names, 
    header=None
)

heartbeat_101 = pd.read_csv(
    "diagnose_TRACE/heartbeat_particle_101.txt", 
    names=heartbeat_column_names, 
    header=None
)

# Filter the first DataFrame
heartbeat_from_begin = heartbeat_embryo_3[heartbeat_embryo_3['t'] >= 83].copy()
heartbeat_101 = heartbeat_101[heartbeat_101['t'] >= 83]
# Drop 'walltime' from both DataFrames in-place
heartbeat_from_begin.drop(columns=['walltime'], inplace=True)
heartbeat_101.drop(columns=['walltime'], inplace=True)
```

```python
t_collision = 8.318553e+01
```

```python
# Ensure they are the same length before comparing
min_length = min(len(heartbeat_from_begin), len(heartbeat_101))
if len(heartbeat_from_begin) != len(heartbeat_101):
    print(f"Warning: Row counts differ ({len(heartbeat_from_begin)} vs {len(heartbeat_101)}). Truncating to {min_length} rows to compare.")
    heartbeat_from_begin = heartbeat_from_begin.iloc[:min_length]
    heartbeat_101 = heartbeat_101.iloc[:min_length]

# FIX: Force exact index alignment (0 to min_length - 1) on both before comparing
heartbeat_from_begin = heartbeat_from_begin.reset_index(drop=True)
heartbeat_101 = heartbeat_101.reset_index(drop=True)

# Compare ONLY the 'energy' column
energy_differences = heartbeat_from_begin['energy'].compare(heartbeat_101['energy'])

if energy_differences.empty:
    print("No differences found in the energy column. The files match perfectly!")
else:
    print(f"Found differences in {len(energy_differences)} rows in the energy column.")
    
    # List the rows where energy values differ
    diff_row_indices = energy_differences.index.tolist()
    print(f"Rows with energy differences: {diff_row_indices}\n")
    energy_differences['delta'] = energy_differences['self'] - energy_differences['other']
    # Display the side-by-side comparison for the energy column
    print("Detailed Energy Differences (self = heartbeat_from_begin, other = heartbeat_101):")
    print(energy_differences)
```
