```python
import rebound
import reboundx
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

pd.set_option('display.precision', 20)
```

```python
heartbeat_column_names = [
    "t", "walltime", "N", "total_mass", 
    "ang_mom_x", "ang_mom_y", "ang_mom_z", "energy", 
    "p3_m", "p3_r", "p3_x", "p3_y", "p3_z", 
    "p3_vx", "p3_vy", "p3_vz"
]

# Read the files
heartbeat_embryo_3 = pd.read_csv(
    "diagnose_TRACE/heartbeat_particle.txt", 
    names=heartbeat_column_names, 
    header=None
)

heartbeat_841 = pd.read_csv(
    "diagnose_TRACE/heartbeat_particle_841.txt", 
    names=heartbeat_column_names, 
    header=None
)

# Filter the first DataFrame
heartbeat_from_begin = heartbeat_embryo_3[heartbeat_embryo_3['t'] > 691.232].copy()

# Drop 'walltime' from both DataFrames in-place
heartbeat_from_begin.drop(columns=['walltime'], inplace=True)
heartbeat_841.drop(columns=['walltime'], inplace=True)
```

```python
# Ensure they are the same length before comparing
min_length = min(len(heartbeat_from_begin), len(heartbeat_841))
if len(heartbeat_from_begin) != len(heartbeat_841):
    print(f"Warning: Row counts differ ({len(heartbeat_from_begin)} vs {len(heartbeat_841)}). Truncating to {min_length} rows to compare.")
    heartbeat_from_begin = heartbeat_from_begin.iloc[:min_length]
    heartbeat_841 = heartbeat_841.iloc[:min_length]

# FIX: Force exact index alignment (0 to min_length - 1) on both before comparing
heartbeat_from_begin = heartbeat_from_begin.reset_index(drop=True)
heartbeat_841 = heartbeat_841.reset_index(drop=True)

# Compare ONLY the 'energy' column
energy_differences = heartbeat_from_begin['energy'].compare(heartbeat_841['energy'])

if energy_differences.empty:
    print("No differences found in the energy column. The files match perfectly!")
else:
    print(f"Found differences in {len(energy_differences)} rows in the energy column.")
    
    # List the rows where energy values differ
    diff_row_indices = energy_differences.index.tolist()
    print(f"Rows with energy differences: {diff_row_indices}\n")
    
    # Display the side-by-side comparison for the energy column
    print("Detailed Energy Differences (self = heartbeat_from_begin, other = heartbeat_841):")
    print(energy_differences)
```

```python
x_differences = heartbeat_from_begin['p3_x'].compare(heartbeat_841['p3_x'])

if x_differences.empty:
    print("No differences found in the X position. The files match perfectly!")
else:
    print(f"Found differences in {len(x_differences)} rows in the X position (p3_x).\n")
    
    # List the rows where X values differ
    diff_row_indices = x_differences.index.tolist()
    print(f"Rows with X differences: {diff_row_indices}\n")
    
    # Add a column showing the exact mathematical difference (delta)
    x_differences['delta'] = x_differences['self'] - x_differences['other']
    
    # Display the side-by-side comparison for the X column
    print("Detailed X Position Differences (Full Precision):")
    print(x_differences)

# Reset pandas display back to default when done
pd.reset_option('display.precision')
```
